```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# --- CONFIGURATION ---
DATA_PATH = r"....." #path to the folder containing the mfcc features extracted using the preprocessing script
BATCH_SIZE = 64
EPOCHS = 300
LEARNING_RATE = 0.001
PATIENCE = 20  

torch.backends.cudnn.benchmark = True
torch.manual_seed(42)
np.random.seed(42)


class MFCC_CNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        
        # Block 1: 
        self.conv1 = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ELU(alpha=1.0, inplace=True),  
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ELU(alpha=1.0, inplace=True),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.1)
        )
        
        # Block 2: 
        self.conv2 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ELU(alpha=1.0, inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ELU(alpha=1.0, inplace=True),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.2)
        )
        
        # Block 3: 
        self.conv3 = nn.Sequential(
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ELU(alpha=1.0, inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ELU(alpha=1.0, inplace=True),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.3)
        )
        
        # Block 4: 
        self.conv4 = nn.Sequential(
            nn.Conv2d(256, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ELU(alpha=1.0, inplace=True),
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Dropout2d(0.4)
        )
        
        # Classifier
        self.fc = nn.Sequential(
            nn.Linear(512, 256),
            nn.ELU(alpha=1.0, inplace=True),
            nn.BatchNorm1d(256),
            nn.Dropout(0.5),
            nn.Linear(256, 128),
            nn.ELU(alpha=1.0, inplace=True),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )
        
        self._init_weights()
    
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0.1)  
            elif isinstance(m, nn.BatchNorm2d) or isinstance(m, nn.BatchNorm1d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.xavier_normal_(m.weight)
                nn.init.constant_(m.bias, 0.1)
    
    def forward(self, x):
        x = self.conv1(x)
        x = self.conv2(x)
        x = self.conv3(x)
        x = self.conv4(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x

class DeadNeuronMonitor:
    def __init__(self, model, threshold=0.95):
        self.model = model
        self.threshold = threshold
        self.hooks = []
        self.activations = {}
        
    def _activation_hook(self, name):
        def hook(module, input, output):
            self.activations[name] = output.detach()
        return hook
    
    def attach_hooks(self):
        self.activations = {}
        for name, module in self.model.named_modules():
            if isinstance(module, (nn.Conv2d, nn.Linear, nn.ELU)):
                self.hooks.append(module.register_forward_hook(self._activation_hook(name)))
    
    def remove_hooks(self):
        for hook in self.hooks:
            hook.remove()
        self.hooks = []
    
    def check_dead_neurons(self, data, device):
        self.model.eval()
        self.attach_hooks()
        with torch.no_grad():
            _ = self.model(data.to(device))
        
        total_dead = 0
        total_neurons = 0
        for name, activation in self.activations.items():
            if isinstance(activation, torch.Tensor):
                if len(activation.shape) == 4:
                    batch, channels, h, w = activation.shape
                    activation = activation.view(batch, channels, -1)
                    inactive = (activation <= 0).float().mean(dim=2)
                    dead = (inactive > self.threshold).sum().item()
                    total_dead += dead
                    total_neurons += channels
                elif len(activation.shape) == 2:
                    batch, features = activation.shape
                    inactive = (activation <= 0).float().mean(dim=0)
                    dead = (inactive > self.threshold).sum().item()
                    total_dead += dead
                    total_neurons += features
        self.remove_hooks()
        return (total_dead / total_neurons) * 100 if total_neurons > 0 else 0.0

def train_model_96(model, train_data, val_data, epochs, device, patience=PATIENCE):
    X_train, y_train = train_data
    X_val, y_val = val_data
    
    train_loader = torch.utils.data.DataLoader(torch.utils.data.TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    val_loader = torch.utils.data.DataLoader(torch.utils.data.TensorDataset(X_val, y_val), batch_size=BATCH_SIZE, shuffle=False)
    
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=LEARNING_RATE * 0.01)
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    monitor = DeadNeuronMonitor(model)
    
    history = {'epoch': [], 'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'dead_train': [], 'dead_val': []}
    
    best_val_acc = 0
    epochs_no_improve = 0  

    print(f"{'Epoch':<6} | {'Train Loss':<10} | {'Val Loss':<10} | {'Train Acc':<10} | {'Val Acc':<10} | {'Dead %':<10}")
    print("-" * 75)
    
    for epoch in range(epochs):
        model.train()
        train_loss, train_correct, train_total = 0, 0, 0
        
        pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}', leave=False)
        for data, target in pbar:
            data, target = data.to(device), target.to(device)
            optimizer.zero_grad()
            output = model(data)
            loss = criterion(output, target)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            
            _, predicted = output.max(1)
            train_total += target.size(0)
            train_correct += predicted.eq(target).sum().item()
            train_loss += loss.item() * target.size(0)
            pbar.set_postfix({'Acc': 100.*train_correct/train_total})
        
        scheduler.step()
        train_loss /= len(X_train)
        train_acc = train_correct / train_total
        
        model.eval()
        val_loss, val_correct, val_total = 0, 0, 0
        with torch.no_grad():
            for data, target in val_loader:
                data, target = data.to(device), target.to(device)
                output = model(data)
                loss = criterion(output, target)
                val_loss += loss.item() * target.size(0)
                _, predicted = output.max(1)
                val_total += target.size(0)
                val_correct += predicted.eq(target).sum().item()
        
        val_loss /= len(X_val)
        val_acc = val_correct / val_total
        
        sample_data, _ = next(iter(train_loader))
        dead_train = monitor.check_dead_neurons(sample_data[:32], device)
        
        history['epoch'].append(epoch + 1)
        history['train_loss'].append(train_loss); history['val_loss'].append(val_loss)
        history['train_acc'].append(train_acc); history['val_acc'].append(val_acc)
        history['dead_train'].append(dead_train)
        
        print(f"{epoch+1:<6} | {train_loss:<10.4f} | {val_loss:<10.4f} | {train_acc:<10.4f} | {val_acc:<10.4f} | {dead_train:<10.1f}")
        
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            epochs_no_improve = 0
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'val_acc': val_acc,
                'history': history
            }, 'best_model.pth')
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"\nEarly stopping triggered after {patience} epochs without improvement.")
                break

    return model, history

def plot_results_table(history):
    print(f"{'Epoch':<6} | {'Train Loss':<12} | {'Val Loss':<12} | "
          f"{'Train Acc':<12} | {'Val Acc':<12} | {'Dead %':<12}")
    print("-" * 90)
    for i in range(0, len(history['epoch']), 10):
        if i < len(history['epoch']):
            idx = i
            print(f"{history['epoch'][idx]:<6} | {history['train_loss'][idx]:<12.4f} | {history['val_loss'][idx]:<12.4f} | "
                  f"{history['train_acc'][idx]:<12.4f} | {history['val_acc'][idx]:<12.4f} | {history['dead_train'][idx]:<12.1f}")
    if len(history['epoch']) > 0:
        idx = -1
        print("-" * 90)
        print(f"{history['epoch'][idx]:<6} | {history['train_loss'][idx]:<12.4f} | {history['val_loss'][idx]:<12.4f} | "
              f"{history['train_acc'][idx]:<12.4f} | {history['val_acc'][idx]:<12.4f} | {history['dead_train'][idx]:<12.1f}")

def plot_training_curves(history):
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    axes[0].plot(history['epoch'], history['train_acc'], label='Train')
    axes[0].plot(history['epoch'], history['val_acc'], label='Val')
    axes[0].axhline(y=0.96, color='r', linestyle='--', label='Target 96%')
    axes[0].set_title('Accuracy'); axes[0].legend(); axes[0].grid(True, alpha=0.3)
    axes[1].plot(history['epoch'], history['train_loss'], label='Train')
    axes[1].plot(history['epoch'], history['val_loss'], label='Val')
    axes[1].set_title('Loss'); axes[1].legend(); axes[1].grid(True, alpha=0.3)
    axes[2].plot(history['epoch'], history['dead_train'], label='Dead %')
    axes[2].set_title('Dead Neurons'); axes[2].grid(True, alpha=0.3)
    plt.tight_layout(); plt.show()

def main():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f" Using device: {device}")
    
    if not os.path.exists(DATA_PATH):
        print(f" Data not found at {DATA_PATH}"); return
    
    data = torch.load(DATA_PATH)
    X_train, y_train = data['X_train'], data['y_train']
    X_val, y_val = data['X_val'], data['y_val']
    
    # Normalization
    mean = X_train.mean(dim=(0, 2, 3), keepdim=True)
    std = X_train.std(dim=(0, 2, 3), keepdim=True) + 1e-8
    X_train = torch.clamp((X_train - mean) / std, -3, 3)
    X_val = torch.clamp((X_val - mean) / std, -3, 3)
    
    model = MFCC_CNN(num_classes=len(torch.unique(y_train))).to(device)
    
   
    model, history = train_model_96(model, (X_train, y_train), (X_val, y_val), epochs=EPOCHS, device=device, patience=PATIENCE)
    
    plot_results_table(history)
    plot_training_curves(history)

    
    checkpoint = torch.load('best_model.pth')
    model.load_state_dict(checkpoint['model_state_dict'])
    
    
    model.eval()
    val_loader = torch.utils.data.DataLoader(torch.utils.data.TensorDataset(X_val, y_val), batch_size=BATCH_SIZE)
    
    all_preds, all_targets = [], []
    with torch.no_grad():
        for data, target in val_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            all_preds.extend(output.argmax(1).cpu().numpy())
            all_targets.extend(target.cpu().numpy())
    
    final_acc = np.mean(np.array(all_preds) == np.array(all_targets))
    print(f"\nBest validation accuracy: {checkpoint['val_acc']:.4f}")
    print(f"Final validation accuracy (restored): {final_acc:.4f}")

    
    torch.save({'model_state_dict': model.state_dict(), 'history': history}, 'mfcc_model_final.pth')

if __name__ == "__main__":
    main()
```
